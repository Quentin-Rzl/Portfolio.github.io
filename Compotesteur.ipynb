```python
def Compotesteur3(arg):
    #VERIFICATION DU NOMBRE D'ARGUMENTS EN ENTREE :
    if len(arg) != 4:
        print("ERREUR : Le nombre de personnages sélectionnés est incorrect !")
        return

    #ATTRIBUTION DES STATISTIQUES :
    
    #Etape 1 : Identification des 3 personnages sélectionnés via liste des noms "Perso_Name", triée par ordre alphabétique.
    #La liste Perso_Num va enregistrer la valeur du nom des arguments sur la liste Perso_Name de chaque argument, par exemple si le 1er argument est
    #"Eni", la première valeur de la liste sera 4.
    
    Perso_Name = ["Cra","Eca","Elio","Eni","Enu","Feca","Iop","Hup","Osa","Ougi","Panda","Roub","Sacri","Sadi","Sram","Steam","Xel","Zobal"]
    Perso_Num = [0,0,0]
    x = 1
    z = 0
    while x < 4 :
        y = 0
        while (Perso_Name[y] != arg[x] and y < 17) :
            y += 1
        if Perso_Name[y] == arg[x] :
            Perso_Num[z] += y
            z += 1
            x += 1
        else :
            print("ERREUR : L'identification d'un personnage à échoué !")
            return
    P1 = Perso_Num[0]
    P2 = Perso_Num[1]
    P3 = Perso_Num[2]
    
    #Etape 2 : Mise en place d'une liste pour chaque statistique comportant le montant de la statistique pour chaque personnage dans l'ordre
    #alphabétique. Les valeurs vont de 1 à 10 afin d'évaluer avec assez de précision chacun des personnages sur chaque statistiques.
    #Les stats sont : les dégats élémentaires (feu, eau, terre et vent), la défense, les soins, les buffs, les débuffs, le placement des alliés et
    #des adversaires.

    Feu = [10,8,0,2,7,2,10,7,4,0,2,10,10,0,10,5,10,6]
    Eau = [0,8,10,8,7,2,0,7,1,10,2,0,0,2,10,10,10,6]
    Terre = [10,8,10,0,7,2,8,7,4,8,2,10,8,6,0,6,0,0]
    Vent = [10,0,10,4,0,0,7,7,4,8,0,10,8,2,10,0,10,6]
    Def = [3,3,3,4,7,10,6,3,4,7,10,3,7,4,3,3,3,4]
    Soin = [2,4,4,10,2,3,2,2,7,2,3,2,2,8,3,4,2,7]
    Buff = [2,6,2,8,2,8,2,2,10,2,6,2,2,6,2,2,2,8]
    Debuff = [2,4,2,8,4,4,4,2,6,4,6,2,2,8,6,2,2,4]
    All = [6,2,10,2,8,4,5,6,2,5,6,5,7,2,8,10,6,6]
    Ene = [3,4,2,3,4,6,2,2,4,4,10,2,7,3,4,3,3,4]

    #Ces listes serviriont de base de données qui sera employé à chacun des calculs suivants en utilisant P1,P2 et P3 comme indice de repérage.

    #DEBUT DE L'ANALYSE :
    
    print("RESULTATS :")
    print(" ")
    print("L'équipe est composée de :",arg[1],";",arg[2],"et",arg[3],"!")
    print(" ")

    #CALCUL DES DEGATS ELEMENTAIRES :

    print("- Dégats élémentaires :")
    print(" ")
    
    #Feu :
    S_Feu = Feu[P1] + Feu[P2] + Feu[P3]
    if S_Feu < 15 :
        print("-> L'équipe n'a pas assez de dégats feu !")
    elif 15 <= S_Feu < 20 :
        print("-> L'équipe a suffisament de dégats feu.")
    elif 20 <= S_Feu :
        print("-> L'équipe a énormément de dégats feu !")
    print("Note :",S_Feu,"/ 30")

    #Eau :
    S_Eau = Eau[P1] + Eau[P2] + Eau[P3]
    if S_Eau < 15 :
        print("-> L'équipe n'a pas assez de dégats eau !")
    elif 15 <= S_Eau < 20 :
        print("-> L'équipe a suffisament de dégats eau.")
    elif 20 <= S_Eau :
        print("-> L'équipe a énormément de dégats eau !")
    print("Note :",S_Eau,"/ 30")
    
    #Terre :
    S_Terre = Terre[P1] + Terre[P2] + Terre[P3]
    if S_Terre < 15 :
        print("-> L'équipe n'a pas assez de dégats terre !")
    elif 15 <= S_Terre < 20 :
        print("-> L'équipe a suffisament de dégats terre.")
    elif 20 <= S_Terre :
        print("-> L'équipe a énormément de dégats terre !")
    print("Note :",S_Terre,"/ 30")
    
    #Vent :
    S_Vent = Vent[P1] + Vent[P2] + Vent[P3]
    if S_Vent < 15 :
        print("-> L'équipe n'a pas assez de dégats vent !")
    elif 15 <= S_Vent < 20 :
        print("-> L'équipe a suffisament de dégats vent.")
    elif 20 <= S_Vent :
        print("-> L'équipe a énormément de dégats vent !")
    print("Note :",S_Vent,"/ 30")
    print(" ")
    
    #CALCUL DEFENSE :
    
    print("- Défense et soins :")
    print(" ")
    
    S_Def = Def[P1] + Def[P2] + Def[P3]
    if S_Def < 13 :
        print("-> L'équipe est trop fragile !")
    elif 13 <= S_Def < 18 :
        print("-> L'équipe est assez résistante.")
    elif 18 <= S_Def :
        print("-> L'équipe à énormément de résistances !")
    print("Note :",S_Def,"/ 20")
    
    #CALCUL SOINS :

    S_Soin = Soin[P1] + Soin[P2] + Soin[P3]
    if S_Soin < 12 :
        print("-> L'équipe ne peut pas se soigner correctement !")
    elif 12 <= S_Soin < 17 :
        print("-> L'équipe a assez de soins.")
    elif 17 <= S_Soin :
        print("-> L'équipe a un surplus de soins !")
    print("Note :",S_Soin,"/ 20")
    print(" ")

    #CALCUL DES BUFFS ET DEBUFFS :

    print("- Buffs et Débuffs :")
    print(" ")

    #Buffs :

    S_Buff = Buff[P1] + Buff[P2] + Buff[P3]
    if S_Buff < 10 :
        print("-> L'équipe a trop peu de buffs !")
    elif 10 <= S_Buff < 15 :
        print("-> L'équipe a assez de buffs.")
    elif 15 < S_Buff :
        print("-> L'équipe a énormément de buffs !")
    print("Note :",S_Buff,"/ 20")

    #Débuffs :

    S_Debuff = Debuff[P1] + Debuff[P2] + Debuff[P3]
    if S_Debuff < 10 :
        print("-> L'équipe a trop peu de débuffs !")
    elif 10 <= S_Debuff < 15 :
        print("-> L'équipe a assez de débuffs.")
    elif 15 < S_Debuff :
        print("-> L'équipe a énormément de débuffs !")
    print("Note :",S_Debuff,"/ 20")
    print(" ")

    #CALCUL DU POSITIONNEMENT ALLIE ET ENNEMI :

    print("- Placement des unités :")
    print(" ")

    #Placement alliés :

    S_All = All[P1] + All[P2] + All[P3]
    if S_All < 15 :
        print("> Alliés : Difficile")
    elif 15 <= S_All < 25 :
        print("> Alliés : Normal")
    elif 25 <= S_All :
        print("> Alliés : Facile")

    #Placement ennemi :

    S_Ene = Ene[P1] + Ene[P2] + Ene[P3]
    if S_Ene < 10 :
        print("> Ennemis : Difficile")
    elif 10 <= S_Ene < 16 :
        print("> Ennemis : Normal")
    elif 16 <= S_Ene :
        print("> Ennemis : Facile")

    print(" ")

    #CALCUL DE L'ORDRE DE JEU :
    #Pour définir l'ordre de jeu entre les personnages, on crée une nouvelle liste "Ordre" dans laquelle on indique les noms des arguments dans
    #l'ordre de passage souhaité du premier au dernier.
    #La liste Perso_Ordre va enregistrer la valeur d'ordre de passage sur la liste Ordre de chaque argument, par exemple si le 1er argument est "Osa",
    #la première valeur de la liste sera 3.

    Ordre = ["Panda","Feca","Osa","Sadi","Zobal","Eni","Eca","Sacri","Enu","Sram","Iop","Ougi","Elio","Hup","Roub","Cra","Xel","Steam"]
    Perso_Ordre = [0,0,0]
    x = 1
    z = 0
    while x < 4 :
        y = 0
        while (Ordre[y] != arg[x] and y < 17) :
            y += 1
        if Ordre[y] == arg[x] :
            Perso_Ordre[z] += y
            z += 1
            x += 1
        else :
            print("ERREUR : L'ordre des personnages n'a pas pu être calculé !")
            return

    #On nomme les valeurs de la liste Perso_Ordre (T1, T2 et T3) afin de pouvoir les rappeler plus facilement et les comparer entre elles.
    
    T1 = Perso_Ordre[0]
    T2 = Perso_Ordre[1]
    T3 = Perso_Ordre[2]

    if T1 <= T2 and T1 <= T3 :
        if T2 <= T3 :
            print("- Ordre de jeu conseillé :",Ordre[T1],">",Ordre[T2],">",Ordre[T3])
        elif T2 > T3 :
            print("- Ordre de jeu conseillé :",Ordre[T1],">",Ordre[T3],">",Ordre[T2])
    elif T2 <= T1 and T2 <= T3 :
        if T1 <= T3 :
            print("- Ordre de jeu conseillé :",Ordre[T2],">",Ordre[T1],">",Ordre[T3])
        elif T1 > T3 :
            print("- Ordre de jeu conseillé :",Ordre[T2],">",Ordre[T3],">",Ordre[T1])
    elif T3 <= T1 and T3 <= T2 :
        if T1 <= T2 :
            print("- Ordre de jeu conseillé :",Ordre[T3],">",Ordre[T1],">",Ordre[T2])
        elif T1 > T2 :
            print("- Ordre de jeu conseillé :",Ordre[T3],">",Ordre[T2],">",Ordre[T1])
    else :
            print("ERREUR : L'ordre des personnages n'a pas pu être calculé !")
            return
    
Compotesteur3([1,"Cra","Sacri","Zobal"])
```
